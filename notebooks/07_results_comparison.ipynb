### honest interpretation

look at the alpha value reported in `hybrid.json`. if it's close to 1.0, the music-theory features didn't add much on this evaluation. if it's lower, they pulled rank. either result is a real finding for an undergrad portfolio project: showing the comparison and saying 'this is what i measured' is better than hiding the negative case.


### where the hybrid is most likely to help

stylistically narrow playlists (a single mood, similar tempo). it is likely to **hurt** on highly eclectic playlists where the user intentionally jumps between styles.
